# Extract the test PDFs

Runs the two PDFs in `tests/test_data`, in filename order, through `PdfExtractor` and `to_markdown`. Figures stay as `<!-- image -->` placeholders. This notebook does not call a vision model.

The first PDF is the 47-page Crop Prospects report. The second is the 13-page `cd9804en.pdf`. The first run also downloads Docling models, so the first extraction takes a while.

Select the project environment as the kernel, then run the cells from top to bottom.

```python
import re
from pathlib import Path

from IPython.display import Markdown, display

from pdf_extractor_ai import PdfExtractor, document_title, is_low_text, split_pages, to_markdown


def repo_root() -> Path:
    start = Path.cwd().resolve()
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").is_file() and (candidate / "tests" / "test_data").is_dir():
            return candidate
    raise FileNotFoundError(f"Could not find the repo root from {start}")


def show_extraction(extractor: PdfExtractor, path: Path) -> None:
    document = extractor.extract(path.read_bytes(), name=path.name)
    markdown = to_markdown(document)
    pages = split_pages(markdown)
    placeholders = len(re.findall(r"<!--\s*image\s*-->", markdown))
    print(path.name)
    print(f"title: {document_title(document, fallback=path.name)}")
    print(f"pages: {len(pages)}")
    print(f"characters: {len(markdown)}")
    print(f"low text: {is_low_text(markdown)}")
    print(f"image placeholders: {placeholders}")
    display(Markdown(markdown))
```

```python
pdfs = sorted((repo_root() / "tests" / "test_data").glob("*.pdf"))
if len(pdfs) < 2:
    raise FileNotFoundError(f"Expected two PDFs in test_data, found {len(pdfs)}")
first_pdf, second_pdf = pdfs[0], pdfs[1]
print("1.", first_pdf.name)
print("2.", second_pdf.name)
```

```python
extractor = PdfExtractor(num_workers=1)
```

## First PDF

```python
show_extraction(extractor, first_pdf)
```

## Second PDF

```python
show_extraction(extractor, second_pdf)
```
